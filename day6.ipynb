**Inheritance**

```python
class Address:
  def __init__(self,city,state):
    self.city=city
    self.state=state
  def location(self):
    return f"my location is {self.city} in {self.state}"

class user(Address):
  def __init__(self,name,age,city,state):
    self.name=name
    self.age=age
    self.city=city
    self.state=state
  def userName(self):
    print(f"my name is {self.name} and my location is {self.city} in {self.state}")
u=user("dheeraj",17,"jaipur","Rajasthan")
print(u.location())
u.userName()
```

**using Super keyword**

```python
class Address:
  def __init__(self,city,state):
    self.city=city
    self.state=state
  def location(self):
    return f"my location is {self.city} in {self.state}"

class user(Address):
  def __init__(self,name,age,city,state):

    super().__init__(city,state)# super using to reduce redeclarartion of variable of parent class
    self.name=name
    self.age=age
  def userName(self):
    print(f"my name is {self.name} and my location is {self.city} in {self.state}")
u=user("dheeraj",17,"jaipur","Rajasthan")
print(u.location())
u.userName()
```

**Encapsulation**

```python
class address:
  def __init__(self,city,state):
    self.__state=state
    self.__city=city
  def location(self):
    print(f"my location is {self.__city} in {self.__state}")
  def getCity(self):
    print(self.__city)
  def getstate(self):
    print(self.__state)
a=address("chhapra","bihar")
a.location()
a.getCity()
a.getstate()
# print(a.__state)
```
