```python
#Recape
class student:
  def __init__(self,age,name):#constructor
    self.age=age
    self.name=name
  def Print(self):
    print(self.age)
    print(self.name)
s=student(18,"Dheeraj")
s.Print()
```

**class Method & self**

```python
class Address:
  def __init__(self, city,state):
    self.city=city
    self.state=state
  def show(self):
    print("the city is :",self.city)
    print("the state is :",self.state)
a=Address("jaipur","rajasthan")
a.show()
```
