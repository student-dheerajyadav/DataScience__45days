**for Loop**

```python
#Example 1.
import numpy as np

arr=np.arange(16).reshape(4,4)
print(arr)
i=0
while i<len(arr):
  j=0
  while j< len(arr[i]):
    if(i==j):
      arr[i][j] = 1
    else:
      arr[i][j]=0
    # print(arr[i][j])
    j+=1
  i += 1
print(arr)
```
