#Handle missing value

```python
 #ENCODING
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
data ={"colors": ['red', 'green', 'yellow', 'green','red',np.nan]}
df=pd.DataFrame(data)
df
#handle null values
df=df.dropna()
print(df)


```

```python

#label Encoding
from sklearn.preprocessing import LabelEncoder
print(df)
le=LabelEncoder()
df["colors_encoder"]=le.fit_transform(df["colors"])
df["colors_encoder1"]=le.fit_transform(df["colors"])
df.drop(["colors_encoder","colors_encoder1"],axis=1,inplace=True)
df
```

**#one Hot Encoder**

```python
import warnings
warnings.filterwarnings('ignore')

from sklearn.preprocessing import OneHotEncoder
import pandas as pd
print(df)
#one hot Encoding
# Create one-hot encoded columns(by the help of sklearn)
encoder=OneHotEncoder()
encoded=encoder.fit_transform(df[["colors"]])
print(encoded.toarray())
# Create one-hot encoded columns(by the help of pandas)
encoded_df = pd.get_dummies(df, columns=['colors'])
print(encoded_df)
```

```python
#combination of handling missing values | label encoding |onehot encoding
import numpy as np
import pandas as pd
from sklearn.preprocessing import label_binarize,OneHotEncoder
data={
    "age":[12,34,np.nan,56,67],
    "name":["dheeraj","kunal","yash","anushka","nikita"],
    "gender":["male","male","male","femle",np.nan]
}
df=pd.DataFrame(data)
print("-------------handling missing values---------")
df["age"]=df["age"].fillna(df["age"].mean())
print(df)
df.dropna(subset=["gender"],inplace=True)
print(df)
print("-----------------label encoding-------------")
lb=LabelEncoder()
df["gender1"]=lb.fit_transform(df["gender"])
print(df)

print("----------------one hot encoding--------------")
on=OneHotEncoder()
on_e=on.fit_transform(df[["gender"]]).toarray()
print(on_e)
```
