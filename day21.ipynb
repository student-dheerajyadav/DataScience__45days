```python
#feature scalling
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
data={
    "experience":[300,600,900,1500],
    "salary":[1000,1500,2000,3000]
}
df=pd.DataFrame(data)
print(df)
#standard scalling
scaler=StandardScaler()
df["experience"]=scaler.fit_transform(df[["experience"]])
print(df)
#split data
X=df["experience"]
Y=df["salary"]
print(X)
print(Y)
plt.title("Data")
plt.xlabel("experience")
plt.ylabel("Salary")
plt.plot(X,Y,marker="o")
plt.grid()
plt.show()
```

```python
#data split training testing
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

url="https://raw.githubusercontent.com/rajendra0968jangid/Ds-Arya/main/extended_salary_data.csv"
df=pd.read_csv(url)
print(df)
#split data
x=df[["experience"]]
y=df[["salary"]]

# train  test Split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=8)
print("------train--------")
print("x_train",x_train)
print("Y_train",y_train)
print("------test------")
print("x_test",x_test)
print("Y_test",y_test)
plt.title("Data")
plt.xlabel("experience")
plt.ylabel("Salary")
plt.plot(X,Y,marker="o")
plt.grid()
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
import pandas as pd

model = LinearRegression()
model.fit(x_train, y_train)
user=int(input("Enter Experience in days"))

new_data = {
    "experience": [user]
}

df1 = pd.DataFrame(new_data)

prediction_data = model.predict(df1)

print("Predicted Salary:", prediction_data[0])
```
