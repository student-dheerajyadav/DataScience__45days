```python
#subplots 1D

import matplotlib.pyplot as plt
year=[2010,2015,2020,2025]
dairy=[100,520,630,400]

#graph 2
year=[1990,2000,2005,2015]
farming=[300,200,250,100]

#subplots
fig,ax=plt.subplots(1,2)
ax[0].plot(year,dairy)
ax[0].set_title("Milk Farms")
ax[0].set_xlabel("year")
ax[0].set_ylabel("dairy")

ax[1].set_title("Farming")
ax[1].set_xlabel("year")
ax[1].set_ylabel("Grains")
ax[1].bar(year,farming)
plt.tight_layout()
plt.show()
```

**2D subplots**

```python
import matplotlib.pyplot as plt
#g raph 1
year=[2010,2015,2020,2025]
dairy=[1000,520,1630,400]
#g raph 2
year=[1990,2000,2005,2015]
farming=[300,200,250,100]

#g raph 3
college=[10,20,30,40]
#g raph 4
transport=[50,75,100,150]

fig,ax=plt.subplots(2,2)

ax[0][0].plot(year,dairy,color="pink",linestyle="--")
ax[0][0].set_title("Dairy")
ax[0][0].set_xlabel("year")
ax[0][0].set_ylabel("cow")
ax[0][1].bar(year,farming,linewidth=3)
ax[0][1].set_title("Farming")
ax[0][1].set_xlabel("year")
ax[0][1].set_ylabel("crop")
ax[1][0].pie(college,labels=year,startangle=90,explode=(0,0,.1,0),shadow=True)
ax[1][0].set_title("DAy scholar")
ax[1][0].set_xlabel("year")
ax[1][0].set_ylabel("student")
ax[1][1].plot(year,transport, marker="o",markersize=10)
ax[1][1].set_title("Transportation")
ax[1][1].set_xlabel("year")
ax[1][1].set_ylabel("vehical")
plt.legend()
plt.tight_layout()
plt.gcf().canvas.get_supported_filetypes()
plt.savefig("subplot.jpg")
plt.show();
```

```python
from re import T
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


url="https://raw.githubusercontent.com/rajendra0968jangid/Ds-arya/main/temperature_data.json"
df=pd.read_json(url)
print(df)
print("-----------------------data cleaning----------------------")
print("---------removal of Nan data in column specific------------- ")
df.dropna(subset=["temperature_c"],inplace=True);
print(df)
print("------------data menage----------")
df.fillna(df["humidity_pct"].mean(),inplace=True)
print(df)
print("--------------Addition of data column in original data--------------")
df["temperature_f"]=df["temperature_c"]*1.8+32
#reset the index
df.reset_index(drop=True,inplace=True)
print(df)

print("-----------plot a graph ---------------")
fig,ax=plt.subplots(1,2)
ax[0].plot(df["temperature_c"],df["humidity_pct"],marker="o",linewidth="4",markerfacecolor='yellow',    # Inner color
         markeredgecolor='red')
ax[0].set_title(" temperature_c vs humidity_pct")
ax[0].set_xlabel("temperature_c")
ax[0].set_ylabel("humidity_pct")
ax[1].pie(df["temperature_f"],labels=df["day"],autopct="%1.1f%%",explode=(0,0,.1,0,0.2,0),shadow=True)
plt.tight_layout()
plt.gcf().canvas.get_supported_filetypes()
plt.savefig("min-project.jpg")
plt.show()

```
