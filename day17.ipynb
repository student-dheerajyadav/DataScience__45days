```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4])
y1 = [30, 20, 20, 40]
y2 = [20, 30, 25, 30]
y3 = [12, 34, 56, 78]

w = 0.25  # width of each bar

plt.bar(x - w, y1, width=w, label="y1")   # shift left
plt.bar(x, y2, width=w, label="y2")       # center
plt.bar(x + w, y3, width=w, label="y3")   # shift right

plt.xlabel("X-axis")
plt.ylabel("Values")
plt.title("Grouped Bar Chart Example")
plt.legend()
plt.show()
```

**pi** *chart*

```python
import matplotlib.pyplot as plt
fruits=["apple","banana","Orange","watermallon"]
count=[40,30,15,78]
plt.plot(fruits,count)
plt.legend()
plt.show()
plt.pie(count,labels=fruits,colors=["red","yellow","Orange","green"],startangle=90,explode=(0,0.1,0,0),shadow=True)
plt.legend()
plt.show()
# plt.plot(count,fruits)
plt.show()
```

**SubPlot**

```python
import matplotlib.pyplot as plt
# first chart
x=[1,2,3,4,5]
y=[10,20,30,40,50]
plt.subplot(1,2,1)# row, column , position
plt.xlabel("x-axis")
plt.ylabel("y-asix")
plt.plot(x,y)

# Second chart
x1=["apple","banana","Orange","watermallon"]
y1=[40,30,15,78]
plt.subplot(1,2,2)
plt.xlabel("x1-axis")
plt.ylabel("y1-axis")
plt.pie(y1,labels=x1)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
# first chart
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4])
Y1 = [30, 20, 20, 40]
Y2 = [20, 30, 25, 30]
Y3 = [12, 34, 56, 78]

w = 0.25  # width of each bar
plt.subplot(2,2,1)
plt.bar(x - w, Y1, width=w, label="y1")   # shift left
plt.bar(x, Y2, width=w, label="y2")       # center
plt.bar(x + w, Y3, width=w, label="y3")   # shift right

plt.xlabel("X-axis")
plt.ylabel("Values")
plt.title("Grouped Bar Chart Example")
plt.legend()

# Second chart
x1=["apple","banana","Orange","watermallon"]
y1=[40,30,15,78]
plt.subplot(2,2,2)
plt.xlabel("x1-axis")
plt.ylabel("y1-axis")
plt.pie(y1,labels=x1)
plt.tight_layout()


#third chart
x2=["red","orange","blue","pink"]
y2=[11,22,33,44]
plt.subplot(2,2,3)
plt.pie(y2,labels=x2)


#fourth chart
x3=[12,1,2,1,42,6]
y3=[12,24,36,48,60,72]
plt.subplot(2,2,4)
plt.xlabel("x-Axis")
plt.ylabel("y-Axix")
plt.plot(x3,y3,marker="o",linestyle="--")

plt.show()
```
