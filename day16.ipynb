**MatplotLib**

**Create first Chart**

```python
import matplotlib.pyplot as plt
x=[2010,2015,2020,2025]
y=[100,200,250,300]
plt.plot(x,y) #lineChart
plt.xlabel("Years") #x-label
plt.ylabel("Sales") # y-label
plt.title("Sales Report")
plt.show()
```

```python
from matplotlib import figure
import matplotlib.pyplot as plt
x=[2010,2015,2020,2025]
y=[100,200,250,300]
plt.plot(x,y)
#graph size
plt.figure(figsize=(6,3)) #figsize(x,y) => x->width , y-> height
plt.plot(x,y)
plt.show()
```

```python
from matplotlib.lines import lineStyles
import matplotlib.pyplot as plt
x=[2010,2015,2020,2025]
y=[100,200,250,300]
# plt.plot(x,y,color="green",marker="o",lineStyles="dashed",linewidth=2,markersize=14)
plt.xlabel("Years")
plt.ylabel("Sales")
plt.title("Sales rate")
plt.plot(x,y,color="blue", marker="o",linestyle="dotted",linewidth="2",markersize=14)
plt.plot(x,y,color="blue", marker="",linestyle="dotted",linewidth="2",markersize=14)
plt.show()
```

**Multiple Line chart**

```python
import matplotlib.pyplot as plt
x=[2010,2015,2020,2025]
y1=[100,200,260,290]
y2=[150,185,195,300]
plt.plot(x,y1,label="jeans",color="blue")
plt.plot(x,y2,label="Shirt",color="pink")
plt.legend() # represt the color line represent which product
plt.show()

```

```python
import matplotlib.pyplot as plt
x=[2010,2015,2020,2025,2030]
y1=[100,200,300,400,500]
y2=[150,250,350,450,550]
y3=[250,350,450,550,650]
plt.xlabel("Year")
plt.ylabel("Sales")
plt.legend()
plt.plot(x, y1,label="Jeans",color="blue", marker="o",linestyle="dotted")
plt.plot(x, y2,label="Shirt",color="red",marker="o",linestyle="dotted",markersize=8)
plt.plot(x, y3,label="Cap",color="brown")
plt.show()
```

**Bar Chart**

```python
import matplotlib.pyplot as plt
x=[2010,2015,2020,2025,2030]
y1=[100,200,300,400,500]
plt.figure( figsize=(6,3))
plt.bar(x,y1,color="red")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array([1,2,3,4])
y1=[30,20,20,40]
y2=[20,30,25,30]
w=0.40
plt.bar(x-w/2,y1,label="Girls",width=w)
plt.bar(x+w/2,y2,label="boys",width=w)
plt.xlabel("Group")
plt.ylabel("no. of student")
plt.title("number of student in ech grp.")
plt.legend()
plt.show()
```
