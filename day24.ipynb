```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Data
df = pd.DataFrame({
    "Experience": [1,2,3,4,5]
})

x = df
y = [10,20,50,90,150]

# Polynomial Features
poly = PolynomialFeatures(degree=2)
x_poly = poly.fit_transform(x)

# Model
model = LinearRegression()
model.fit(x_poly, y)

# Prediction
input_data = int(input("Enter Experience: "))

new_data = pd.DataFrame({
    "Experience": [input_data]
})

new_poly = poly.transform(new_data)   # transform() only

prediction = model.predict(new_poly)

print("Predicted Salary:", prediction[0])

# Graph
y_pred = model.predict(x_poly)

plt.scatter(x["Experience"], y, color="red", label="Actual Data")
plt.plot(x["Experience"], y_pred, color="blue", label="Polynomial Curve")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.legend()
plt.show()
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Data
df = pd.DataFrame({
    "label": [1,2,3,4,5]
})

x = df
y = [10,20,50,90,150]

# Model
model = LinearRegression()
model.fit(x, y)

# Prediction
input_data = int(input("Enter the label: "))

new_data = pd.DataFrame({
    "label": [input_data]
})

pre_D = model.predict(new_data)

print("Prediction:", pre_D[0])

# Prediction on training data
y_pred = model.predict(x)

# Graph
plt.scatter(x["label"], y, color="red", label="Actual Data")
plt.plot(x["label"], y_pred, color="blue", label="Regression Line")
plt.scatter(input_data, pre_D[0], color="green", s=100, label="Predicted Point")

plt.xlabel("Label")
plt.ylabel("Output")
plt.legend()
plt.show()
```

```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "label":[1,2,3,4,5]
})

x = df
y = [10,20,50,90,150]

# Polynomial Features
poly = PolynomialFeatures(degree=3)
x_poly = poly.fit_transform(x)

# Model
model = LinearRegression()
model.fit(x_poly, y)

# Prediction
input_data = int(input("Enter Label: "))

new_data = pd.DataFrame({
    "label":[input_data]
})

new_poly = poly.transform(new_data)

prediction = model.predict(new_poly)

print("Prediction =", prediction[0])

# Graph
y_pred = model.predict(x_poly)

plt.scatter(x["label"], y, color="red", label="Actual")
plt.plot(x["label"], y_pred, color="blue", label="Polynomial Curve")
plt.xlabel("Label")
plt.ylabel("Output")
plt.legend()
plt.show()
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

# Data
df = pd.DataFrame({
    "weight": [120, 140, 150, 170],
    "size": [6, 7, 7, 8]
})

# Features
x = df

# Target
y = ["Apple", "Apple", "Orange", "Orange"]

# Model
model = DecisionTreeClassifier()
model.fit(x, y)

# Prediction Input
input_weight = int(input("Enter the weight: "))
input_size = int(input("Enter the size: "))

new_data = pd.DataFrame({
    "weight": [input_weight],
    "size": [input_size]
})

# Prediction
predicted_data = model.predict(new_data)

print("Predicted Fruit:", predicted_data[0])
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier,plot_tree
import matplotlib.pyplot as plt


# Dataset
df = pd.DataFrame({
    "Marks": [20, 35, 40, 50, 60, 75, 90]
})

# Features
x = df

# Target
y = ["Fail", "Fail", "Pass", "Pass", "Pass", "Pass", "Pass"]

# Model
model = DecisionTreeClassifier()
model.fit(x, y)

# User Input
marks = int(input("Enter your marks: "))

new_data = pd.DataFrame({
    "Marks": [marks]
})

# Prediction
result = model.predict(new_data)

print("Result:", result[0])
# Decision Tree Graph
plt.figure(figsize=(8,5))

plot_tree(
    model,
    feature_names=["Marks"],
    class_names=["Fail", "Pass"],
    filled=True,
    rounded=True
)

plt.show()
```
