```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder,StandardScaler
from sklearn.model_selection import  train_test_split
from sklearn.linear_model import LinearRegression
# !pip install pymysql;
import pymysql
conn=pymysql.connect(host="localhost",user="root",database="electricity_project",password="Dheeraj@123")
print(conn);
query="select units,members,ac_hours,cooler_hours,tv_hours from electricity_bill"
df=pd.read_sql(query,conn)


# feature scaling
scaler=StandardScaler()
df[["units","members","ac_hours","cooler_hours","tv_hours"]]=scaler.fit_transform(df[["units","members","ac_hours","cooler_hours","tv_hours"]])


#split data
x=df[["units","members","ac_hours","cooler_hours","tv_hours"]]
y=df[["bill"]]

#train test split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.2,random_state=42)

#model train
model=LinearRegression()
model.fit(x_train,y_train)
units=int(input("enter units"))
m=int(input("enter no. of members"))
ac=float(input("enter a.c hours"))
c=float(input("enter cooler hours"))
tv=float(input("enter Tv hours"))
new_data =pd.DataFrame({
    "units":[units],
    "members":[m],
    "ac_hours":[ac],
    "cooler_hours":[c],
    "tv_hours":[tv]
})
new_data[["units","members","ac_hours","cooler_hours","tv_hours"]]=scaler.transform(new_data[["units","members","ac_hours","cooler_hours","tv_hours"]])
#  #predict
predict_Data=model.predict(new_data)
print("bill : ",predict_Data[0])

```
